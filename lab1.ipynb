### Class

```python
class test(object):

    def __init__(self, input1, input2):
        self.a = input1
        self.b = input2

    def calculate(self, t1, t2):
        result = t1*self.a + t2*self.b
        return result

test_class = test(input1=2, input2=3)
print(f'a = {test_class.a}')
print(f'compute: {test_class.calculate(t1=1, t2=3)}')
```

### 2. Backward Propogation

Another major feature of PyTorch is the ability to automatically differentiate the calculation process, making it easy to perform backward propagation. Every operation on Tensor leaves a computation graph and preserves the path for differentiation.

#### 2.1 Acquring gradients using Autograd

```python
# First, creat a tensor that enables gradient calculation
a_ = torch.ones((2,2), requires_grad=True)
a_
```

```python
# No operation is performed on a_, so the current gradient is None
gradient = a_.grad
print(gradient)
```

```python
# sum operation
b_ = a_ + 1
b_
```

We can see that there is a grad_fn in tensor b, which records the operation of creating tensor b, making it easier for backpropagation (differentiation) in the future

```python
# Sum the elements in b_
sum_ = b_.sum()
sum_
```

```python
# Perform backward propogation
sum_.backward()
```

```python
# Now we can see the gradient of a_
a_.grad
```

It is worth noting that after calling the ` backward() ` function once, the calculation graph involved will be cleared to zero. If ` backward() ` is called twice in a row, the second ` backward() ` will report an error

### 2.2 Optimizer
Now that I know that the gradient of the tensor can be obtained through the ` backward() ` function, I can directly use gradient descent to update the parameters. There are many variations of gradient descent, such as SGD, Momentum gradient descent, Adam gradient descent. PyTorch has already implemented these optimization methods for us in 'torch. opt', and we can use them directly.

```python
import torch.nn as nn
param = nn.Parameter(torch.ones(2,2))
optimizer0 = torch.optim.SGD([param], lr=0.1)
```

```python
param
```

```python
# Define a loss function that sum all the squared elements in the parameter matrix
loss0 = (param**2).sum()
```

```python
loss0
```

```python
# Before each backward, we need to call zero_grad() to clear the gradient in the parameter, otherwise the gradient value will continue to accumulate
optimizer0.zero_grad()
loss0.backward()
optimizer0.step() # step() means that you will update you parameters using SGD
print(param)
```

```python
(param**2).sum()
```

```python
# An iteration is applied to continue optimizing the parameter
for i in range(20):
    optimizer0.zero_grad()
    loss0 = (param**2).sum()
    loss0.backward()
    optimizer0.step()
    print('loss value: {:.4f}'.format(loss0.item()))
print('Optimized：', param)
```

From the above example, we can see that the 'loss' value is continuously decreasing, and in the end, we obtain a 'param' that is very close to 0.
